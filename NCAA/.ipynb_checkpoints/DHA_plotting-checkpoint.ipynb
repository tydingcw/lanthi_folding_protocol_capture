```python
def plot_rama(rama_list, title='Dehydroalanine Ramachandran Plot', dim1 = 15, dim2 = 15):
    prob_matrix=[]
    num_boxes=37
    for i in range(num_boxes):
        prob_matrix.append([])
        for j in range(num_boxes):
            prob_matrix[i].append(0)
    
    #max_amino = 0
    #num_amino = len(rama_list)
    diff = 360/(37-1)
    for pair in rama_list:
        phi=int((pair[0]+180+diff/2)/diff)
        psi=36-int((pair[1]+180+diff/2)/diff)
        if phi == 36:
        #    print('phi 36')
            phi = 0
        if psi == 36:
        #    print('psi 36')
            psi = 0
        prob_matrix[psi][phi] = pair[2]
        #accounting for -180 and 180 being the same value
        if phi == 0 and psi == 0:
            #print('doub 0', pair[0], pair[1])
            prob_matrix[36][36] = pair[2]
            prob_matrix[0][36] = pair[2]
            prob_matrix[36][0] = pair[2]
        elif psi == 0:
            #print('psi 0', pair[0], pair[1])
            prob_matrix[36][phi] = pair[2]
        elif phi == 0:
            #print('phi 0', pair[0], pair[1])
            prob_matrix[psi][36] = pair[2]
        #if prob_matrix[psi][phi] > max_amino:
        #    max_amino += 1
    
    #for i in range(num_boxes):
    #    for j in range(num_boxes):
    #        #prob_matrix[i][j] /= max_disulfides
    #        prob_matrix[i][j] /= num_amino
    
    #print(max_disulfides)
    #print(num_disulfides)
    
    plt.rcParams['font.size'] = 30
    plt.rcParams['axes.labelsize'] = 30
    plt.rcParams['xtick.labelsize'] = 24
    plt.rcParams['ytick.labelsize'] = 24
    plt.rcParams['svg.fonttype'] = 'none'
    
    tick_marks = [x for x in range(num_boxes)]
    xtarget_names=[int(-180 + 360*(x/(num_boxes-1))) for x in range(num_boxes)]
    ytarget_names=[int(180 - 360*(x/(num_boxes-1))) for x in range(num_boxes)]
    
    
    cmap = plt.get_cmap('YlOrRd')

    plt.figure(figsize=(dim1, dim2))
    
    ax=plt.subplot(111)
    #plt.imshow(prob_matrix, cmap=cmap, norm=colors.LogNorm(vmin=0.00001, vmax=1.0))
    #prob_matrix[28][8] = 0.015
    plt.imshow(prob_matrix, cmap=cmap)
    plt.colorbar(label='Probability')
    plt.title(title)
    plt.ylabel('Psi Dihedral')
    plt.xlabel('Phi Dihedral')
    plt.yticks(tick_marks, ytarget_names)
    plt.xticks(tick_marks, xtarget_names)
    ax.set_xticks([x*3 for x in range(13)])
    ax.set_yticks([x*3 for x in range(13)])
    plt.show()
    
```

```python
def read_rama(filename, title, prnt=False):
    rama_list = []
    max_prob = [0, 0, 0]
    with open(filename, 'r') as infile:
        for raw_line in infile:
            if '#' != raw_line[0]:
                line = raw_line.strip()
                split = line.split()
                phi = int(split[1])
                psi = int(split[2])
                prob = float(split[3])
                #rama_list.append((phi, psi, min(0.01, prob)))
                rama_list.append((phi, psi, prob))
                if prob > max_prob[2]:
                    max_prob = [phi, psi, prob]

    plot_rama(rama_list, title)
    
    if prnt:
        print(max_prob)
```

```python
#Averaging of DBU alpha and beta
#psi 40 to -50
#phi -30 to -140
rama_dict = {}
rama_dict['alpha'] = {}
rama_dict['beta'] = {}
beta_tot = 0.0
alpha_tot = 0.0

out_dict = {}
for i in range (-180, 180, 10):
    out_dict[i] = {}
    #for j in range(-180, 190, 10):
    #    out_dict[i]
#print(out_dict)

beta_list = []
alpha_list = []
with open(dbu_beta_path, 'r') as infile:
    for raw_line in infile:
        if '#' != raw_line[0]:
            line = raw_line.strip()
            split = line.split()
            phi = int(split[1])
            psi = int(split[2])
            prob = float(split[3])
            if psi < 41 and psi > -51 and phi < -31 and phi > -141:
                rama_dict['beta'][(phi, psi)] = prob
                beta_tot += prob
            else:
                beta_list.append((phi, psi, prob))
                out_dict[phi][psi] = prob
with open(dbu_alpha_path, 'r') as infile:
    for raw_line in infile:
        if '#' != raw_line[0]:
            line = raw_line.strip()
            split = line.split()
            phi = int(split[1])
            psi = int(split[2])
            prob = float(split[3])
            if psi < 41 and psi > -51 and phi < -31 and phi > -141:
                rama_dict['alpha'][(phi, psi)] = prob
                alpha_tot += prob
                #print(line)
            else:
                alpha_list.append((phi, psi, prob))

#need to normalize alpha values
#print(beta_tot, alpha_tot) # 0.5291082972541252 0.9455407448496287
for key in rama_dict['alpha'].keys():
    rama_dict['alpha'][key] = rama_dict['alpha'][key] * (beta_tot/alpha_tot)

#now, reweight beta
beta_weight = 0.5
for key in rama_dict['alpha'].keys():
    prob = rama_dict['beta'][key] * beta_weight + rama_dict['alpha'][key] * (1-beta_weight)
    beta_list.append((key[0], key[1], prob))
    out_dict[key[0]][key[1]] = prob
            
plot_rama(beta_list, 'DBU Reweighted')

tot_prob = 0.0
for item in beta_list:
    tot_prob += item[2]
#print(tot_prob) #0.9999999999999999

str_out = '''#Resname phi psi prob -log(prob)
''' #natural log
#print(out_dict[-180])
for phi in range (-180, 180, 10):
    for psi in range (-180, 180, 10):
        #print(phi, psi)
        str_out += f'DBU {phi} {psi} {out_dict[phi][psi]} {np.log(out_dict[phi][psi])}\n'
#print(str_out)

out_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/NCAA/DBU_rosetta/params/DBU.rama'
with open(out_path, 'w') as out_file:
    out_file.write(str_out)
```

```python
def parse_cmap(filename):
    cmap_data = []
    with open(filename, 'r') as file:
        lines = file.readlines()
        # Find the start of the CMAP data
        start_index = lines.index("%FLAG CMAP_PARAMETER\n")  # Skip the "%FORMAT(...)" line
        #print(start_index)
        #print(lines[start_index])
        # Read the CMAP data
        for i in range(start_index+1, start_index + 72+1):  # 72 rows of CMAP data
            values = lines[i].split()
            cmap_data.extend([float(v) for v in values])
            # Reverse the order of values in each row and convert to floats
            #reversed_values = [float(v) for v in values[::-1]]
            #cmap_data.extend(reversed_values)
    # Reshape the flat list into a 24x24 grid
    #cmap_grid = [cmap_data[i:i+24] for i in range(0, len(cmap_data), 24)]
    #cmap_grid.reverse()  # Reverse the rows to start from -180 degrees
    cmap_grid = np.array(cmap_data).reshape((24, 24))
    cmap_grid = np.rot90(cmap_grid)
    #cmap_grid = np.flip(cmap_grid, axis=0)  # Flip the rows
    #cmap_grid = np.flip(cmap_grid, axis=1)  # Flip the columns
    #print(cmap_grid)
    #print(len(cmap_grid), len(cmap_grid[-1]))
    return cmap_grid
```

```python
def parse_cmap(filename, instance=1):
    cmap_data = []
    with open(filename, 'r') as file:
        lines = file.readlines()
        # Find the start of the CMAP data for the specified instance
        start_indices = [i for i, line in enumerate(lines) if line.startswith("%FLAG CMAP_PARAMETER")]
        #print(start_indices)
        #print(len(start_indices))
        if instance > len(start_indices):
            raise ValueError(f"Instance {instance} of '%FLAG CMAP_PARAMETER' not found in file.")
        start_index = start_indices[instance - 1]  # Adjust for zero-based indexing
        
        # Read the CMAP data
        for i in range(start_index + 1, start_index + 1 + 72):  # 72 rows of CMAP data
            values = lines[i].split()
            #print(values)
            cmap_data.extend([float(v) for v in values])
    
    # Reshape the flat list into a 24x24 grid and rotate 90 degrees counterclockwise
    cmap_grid = np.array(cmap_data).reshape((24, 24))
    cmap_grid = np.rot90(cmap_grid)
    
    return cmap_grid
```

```python
def plot_cmap(cmap_grid, title='CMAP Ramachandran Plot', dim1=15, dim2=15):
    # Convert the CMAP grid to a NumPy array for easier plotting
    prob_matrix = np.array(cmap_grid)

    plt.rcParams['font.size'] = 30
    plt.rcParams['axes.labelsize'] = 30
    plt.rcParams['xtick.labelsize'] = 24
    plt.rcParams['ytick.labelsize'] = 24
    plt.rcParams['svg.fonttype'] = 'none'
    
    num_boxes = prob_matrix.shape[0]  # Should be 24 for a 24x24 CMAP grid
    tick_marks = np.arange(num_boxes)
    xtarget_names = np.linspace(-180, 180, num_boxes)
    ytarget_names = np.linspace(180, -180, num_boxes)
    
    #cmap = plt.get_cmap('YlOrRd')
    cmap = plt.get_cmap('bwr')
    plt.figure(figsize=(dim1, dim2))
    
    ax = plt.subplot(111)
    plt.imshow(prob_matrix, cmap=cmap)
    plt.colorbar(label='CMAP Value')
    plt.title(title)
    plt.ylabel('Psi Dihedral')
    plt.xlabel('Phi Dihedral')
    plt.yticks(tick_marks, ytarget_names)
    plt.xticks(tick_marks, xtarget_names)
    ax.set_xticks(np.linspace(0, num_boxes-1, 13))
    ax.set_yticks(np.linspace(0, num_boxes-1, 13))
    plt.show()
```
