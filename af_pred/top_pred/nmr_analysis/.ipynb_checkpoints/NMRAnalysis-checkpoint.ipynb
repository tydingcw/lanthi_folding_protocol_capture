```python
def get_min_scoring(scorefile):
    cols = None
    with open(scorefile, 'r') as in_file:
        first_line = in_file.readline()
        first_line = in_file.readline()
        cols=first_line.split()
    
    df = pd.read_csv(scorefile, delim_whitespace=True, skiprows=1)[cols]
    return df.nsmallest(n=1, columns=['total_score'])['description'].iloc[0]
```

```python
#How to analyze NMR constraints
#Do this on the per residue level
#First read in the constraint file, determine cst order (dict) and residue pair
#initialize a grid (nxn residues) with cst value of -1 (no cst)
#for the selected structure, if cst == -1, set to value, else add
# if the pair is not the same residue, do this for the alternative numbering as well. 

def truncate_colormap(cmap, minval=0.0, maxval=1.0, n=100):
    new_cmap = colors.LinearSegmentedColormap.from_list(
        'trunc({n},{a:.2f},{b:.2f})'.format(n=cmap.name, a=minval, b=maxval),
        cmap(np.linspace(minval, maxval, n)))
    return new_cmap

def gen_nmr_plot(name, cst_file, score_file, pdb):
    #first get the length of the peptide
    #struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/mem_try2/'
    struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/'
    cmd.reinitialize()
    cmd.load(f'{struct_path}{pdb}/{name}.pdb.gz')
    stored.holder = []
    cmd.select('pep_sele', 'chain A')
    cmd.iterate( 'pep_sele', 'stored.holder.append( (chain, resi))')
    num_resi = len(set(stored.holder))
    
    blank = -2.0
    matrix = []
    for i in range(num_resi):
        matrix.append([])
        for j in range(num_resi):
            matrix[i].append(blank)
    
    #print(matrix)
    
    cst_dict = {}
    #file with nmr constraints
    with open(struct_path+cst_file, 'r') as in_file:
        for raw_line in in_file:
            line = raw_line.strip()
            split = line.split() #2 and 4 have resi number, -1 has cst number, 1 and 3 have atom
            cst_dict[split[-1]] = (int(split[2]), int(split[4]), split[1], split[3])
    
    #CstFile1_100_score
    #file with score for constraints
    with open(struct_path+score_file, 'r') as in_file:
        first_line = in_file.readline()
        first_line = in_file.readline()
        cols=first_line.split()
    
    df = pd.read_csv(struct_path+score_file, delim_whitespace=True, skiprows=1)[cols]
    df = df.loc[df['description'] == name ]
    
    for col in cols:
        if 'score' in col and 'CstFile' in col:
            #print(col)
            num = col.split('_')[1]
            pair = cst_dict[num]
            i = pair[0] -1
            j = pair[1] -1
            score = float(df[col])
            #print(matrix[i])
            #print(i, j, matrix[i][j], score)
            #print(matrix[i][j] + score)
            if i == j:
                if matrix[i][j] == blank:
                    matrix[i][j] = 0.0
                matrix[i][j] += score
            else:
                if matrix[i][j] == blank:
                    matrix[i][j] = 0.0
                    matrix[j][i] = 0.0
                matrix[i][j] += score
                matrix[j][i] += score
            #print(matrix[i][j])
            if matrix[i][j] > 0.0:
                #print constraint here too
                print(i+1, pair[2], j+1, pair[3], score, matrix[i][j])
    
    plt.figure(figsize=(8, 8))
    cmap=plt.get_cmap('bwr')
    magnitude = max(np.max(matrix), abs(np.min(matrix)))
    cmap = truncate_colormap(cmap, 0.0, 1.0)
    plt.imshow(matrix, cmap=cmap, vmin=-magnitude, vmax=magnitude)
    plt.xlabel('Residue')
    plt.ylabel('Residue')
    plt.title(f'NMR Constraint Deviation {pdb}')
    plt.colorbar()
    #if pdb == '2KTO':
    #    num_resi -= 1
    y_names = [x+1 for x in range(num_resi)]
    y_ticks = [x for x in range(num_resi)]
    x_names = [x+1 for x in range(num_resi)]
    x_ticks = [x for x in range(num_resi)]
    plt.yticks(y_ticks, y_names)
    plt.xticks(x_ticks, x_names)
    plt.show()
```

# NMR Struct Analysis

```python
def calc_nmr_clust_dev(score_file, struct_list = []):
    #potentially expand to examine variable in individual constraints as well
    struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/'
    
    #CstFile1_100_score
    #file with score for constraints
    with open(struct_path+score_file, 'r') as in_file:
        first_line = in_file.readline()
        first_line = in_file.readline()
        cols=first_line.split()
    
    df = pd.read_csv(struct_path+score_file, delim_whitespace=True, skiprows=1)[cols]
    #df = df.loc[df['description'] == name ]
    
    for column in df.columns:
        if 'score' in column or 'measure' in column:
            df[column] = pd.to_numeric(df[column], errors='coerce')
    
    #print(df)
    
    if struct_list:
        df = df[df['description'].isin(struct_list)]
    
    #iterate through all rows. Calculate a deviation per structure
    #total_score = 0
    score_list = []
    struct = 1
    for index, row in df.iterrows():
        total_score = 0
        for column_name, value in row.items():
            if 'score' in column_name:
                #print(column_name, value)
                #if value != 0.0:
                #    print(column_name, value)
                total_score += value
        print(f'struct {struct} score: {total_score}')
        score_list.append(total_score)
        struct += 1
    #print(score_list) 
    print(np.mean(score_list), round(np.std(score_list), 3))
    
    #iterate through all score columns, take the min and compute ensemble deviation
    ensemble_score = 0
    for column_name in df.columns:
        if 'score' in column_name:
            column_min = df[column_name].min()
            #print(column_name, column_min)
            ensemble_score += column_min
    print(f'ensemble score: {ensemble_score}')
```

# Analysis of Rosetta NMR Constraint Ensembles

```python
# Want to calculate the average score and the deviation of the scores
#gen_nmr_plot(name, cst_file, score_file, pdb)
#cst_file has the list of constraints
#name used to load pdb and get length
#pdb is redundand 4 letter code
#score file has cst values and scores
#input should be cst_file, score_file, ensemble_file (name and pdb too?)
def gen_ensemble_plot(name, cst_file, score_file, ensemble_file, pdb):
    #first get the length of the peptide
    #struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/mem_try2/'
    struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/'
    cmd.reinitialize()
    cmd.load(f'{struct_path}{pdb}/{name}.pdb.gz')
    stored.holder = []
    cmd.select('pep_sele', 'chain A')
    cmd.iterate( 'pep_sele', 'stored.holder.append( (chain, resi))')
    num_resi = len(set(stored.holder))
    
    #sets the matrix to default to -2.0 
    blank = -1.0
    matrix = []
    mean_matrix = []
    std_matrix = []
    for i in range(num_resi):
        matrix.append([])
        mean_matrix.append([])
        std_matrix.append([])
        for j in range(num_resi):
            matrix[i].append(blank)
            mean_matrix[i].append(blank)
            std_matrix[i].append(blank)
    
    #print(matrix)
    
    #what are the nmr constraints (residue and atom)?
    cst_dict = {}
    #file with nmr constraints
    with open(struct_path+cst_file, 'r') as in_file:
        for raw_line in in_file:
            line = raw_line.strip()
            split = line.split() #2 and 4 have resi number, -1 has cst number, 1 and 3 have atom
            cst_dict[split[-1]] = (int(split[2]), int(split[4]), split[1], split[3])
    
    #Which structures do we care about (ensemble)
    ens_structs = []
    with open(struct_path+ensemble_file, 'r') as in_file:
        for raw_line in in_file:
            line = raw_line.strip()
            if 'score' not in line:
                ens_structs.append(line.split('/')[-1].split('.')[0])
    
    #initialize the columns
    #CstFile1_100_score
    #file with score for constraints
    with open(struct_path+score_file, 'r') as in_file:
        first_line = in_file.readline()
        first_line = in_file.readline()
        cols=first_line.split()
    
    for name in ens_structs:
        #for each file in ensemble, get the constraints
        df = pd.read_csv(struct_path+score_file, delim_whitespace=True, skiprows=1)[cols]
        df = df.loc[df['description'] == name ]

        for col in cols:
            if 'score' in col and 'CstFile' in col:
            #if 'measure' in col and 'CstFile' in col:
                #print(col)
                num = col.split('_')[1]
                pair = cst_dict[num]
                i = pair[0] -1
                j = pair[1] -1
                score = float(df[col])
                #print(matrix[i])
                #print(i, j, matrix[i][j], score)
                #print(matrix[i][j] + score)
                if i == j:
                    if matrix[i][j] == blank:
                        matrix[i][j] = []
                    matrix[i][j].append(score)
                else:
                    if matrix[i][j] == blank:
                        matrix[i][j] = []
                        matrix[j][i] = []
                    matrix[i][j].append(score)
                    matrix[j][i].append(score)
                #print(matrix[i][j])
                #if matrix[i][j] > 0.0:
                    #print constraint here too
                    #print(i+1, pair[2], j+1, pair[3], score, matrix[i][j])
    
    #process the cst information (mean and std dev)
    #print(matrix)
    #print('****************')
    #print(np.mean(matrix))
    #print('****************')
    for i in range(num_resi):
        for j in range(num_resi):
            mean_matrix[i][j] = np.mean(matrix[i][j])
            std_matrix[i][j] = np.std(matrix[i][j])
    
    #plotting
    plt.figure(figsize=(8, 8))
    cmap=plt.get_cmap('bwr')
    magnitude = max(np.max(mean_matrix), abs(np.min(mean_matrix)))
    cmap = truncate_colormap(cmap, 0.0, 1.0)
    plt.imshow(mean_matrix, cmap=cmap, vmin=-magnitude, vmax=magnitude)
    plt.xlabel('Residue')
    plt.ylabel('Residue')
    plt.title(f'NMR Mean Constraint Deviation {pdb}')
    plt.colorbar()
    #if pdb == '2KTO':
    #    num_resi -= 1
    y_names = [x+1 for x in range(num_resi)]
    y_ticks = [x for x in range(num_resi)]
    x_names = [x+1 for x in range(num_resi)]
    x_ticks = [x for x in range(num_resi)]
    plt.yticks(y_ticks, y_names)
    plt.xticks(x_ticks, x_names)
    plt.show()
    
    #plotting
    plt.figure(figsize=(8, 8))
    cmap=plt.get_cmap('bwr')
    magnitude = max(np.max(std_matrix), abs(np.min(std_matrix)))
    cmap = truncate_colormap(cmap, 0.0, 1.0)
    plt.imshow(std_matrix, cmap=cmap, vmin=-magnitude, vmax=magnitude)
    plt.xlabel('Residue')
    plt.ylabel('Residue')
    plt.title(f'NMR STD Constraint Deviation {pdb}')
    plt.colorbar()
    #if pdb == '2KTO':
    #    num_resi -= 1
    y_names = [x+1 for x in range(num_resi)]
    y_ticks = [x for x in range(num_resi)]
    x_names = [x+1 for x in range(num_resi)]
    x_ticks = [x for x in range(num_resi)]
    plt.yticks(y_ticks, y_names)
    plt.xticks(x_ticks, x_names)
    plt.show()
```

```python
#gets the center distances from the nmr
def parse_reference(filename):
        ref_dict = {}
        aa_list = ['ALA', 'ARG', 'ASN', 'ASP', 'CYS', 'GLN', 'GLU', 'GLY', 'HIS', 'ILE',
           'LEU', 'LYS', 'MET', 'PHE', 'PRO', 'SER', 'THR', 'TRP', 'TYR', 'VAL',
           'DBB', 'DBU', 'DHA', 'DBS', 'DBR', 'DAL', 'ABA'] #'BDS', 'BLR'
        with open (filename, 'r') as infile:
                count = 0
                for raw_line in infile:
                        line = raw_line.strip()
                        split = line.split()
                        hash_split = line.split('#')[0].split()
                        if len(line) == 0 or line[0] == '#':
                                pass
                        elif line[0:6] == 'assign': #this is for XPLOR type files
                                center = float(split[11])
                                if center > 2.3 :
                                        count += 1
                                        ref_dict[count] = center
                        elif len(split) == 64 and split[8] in aa_list and split[18] in aa_list: #2KTO
                                center = float(split[28])
                                if center > 2.3 :
                                        #temp_str = make_constraint(res1, res2, atom1, atom2, center, lower, upper)
                                        count += 1
                                        ref_dict[count] = center
                                        #out_str += temp_str.replace('count', str(count))
                        elif len(hash_split) == 7 and split[1] in aa_list and split[4] in aa_list:  # 6PQG
                                center = float(split[6])
                                if center > 2.3:
                                        count += 1
                                        ref_dict[count] = center
                        else:
                                print(f'{split[8]} or {split[18]} not an AA')
        return ref_dict
```

```python
def get_ensemble_list(filename):
    ens_structs = []
    with open(filename, 'r') as in_file:
        for raw_line in in_file:
            line = raw_line.strip()
            if 'score' not in line:
                ens_structs.append(line.split('/')[-1].split('.')[0])
    return ens_structs
```

```python
# Want to calculate the average score and the deviation of the scores
#gen_nmr_plot(name, cst_file, score_file, pdb)
#cst_file has the list of constraints
#name used to load pdb and get length
#pdb is redundand 4 letter code
#score file has cst values and scores
#input should be cst_file, score_file, ensemble_file (name and pdb too?)
def gen_ensemble_noe_plot(example, pdb, nmr=False, structure='helix'):
    if structure == 'helix':
        struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/'
    elif structure == 'unstructured':
        struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/conf_gen/multi_ripp_scan/xml_test/'
    else:
        print(f'{structure} is not a structure option')
        exit()
    cst_file = pdb + '_noe.cst'
    if nmr:
        score_file = pdb + '_cst_nmr.sc'
    else:
        score_file = pdb + '_cst.sc'
    ensemble_file =  pdb + '_ensemble.txt'
    ref_cst = struct_path + pdb + '_noe.tbl'
    reference = parse_reference(ref_cst)
    #first get the length of the peptide
    #struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/mem_try2/'
    cmd.reinitialize()
    cmd.load(f'{struct_path}{pdb}/{example}.pdb.gz')
    stored.holder = []
    cmd.select('pep_sele', 'chain A')
    cmd.iterate( 'pep_sele', 'stored.holder.append( (chain, resi))')
    num_resi = len(set(stored.holder))
    
    #sets the matrix to default
    blank = -1.0
    matrix = []
    mean_matrix = []
    std_matrix = []
    for i in range(num_resi):
        matrix.append([])
        mean_matrix.append([])
        std_matrix.append([])
        for j in range(num_resi):
            matrix[i].append(blank)
            mean_matrix[i].append(blank)
            std_matrix[i].append(blank)
    
    #what are the nmr constraints (residue and atom)?
    cst_dict = {}
    #file with nmr constraints
    with open(struct_path+cst_file, 'r') as in_file:
        for raw_line in in_file:
            line = raw_line.strip()
            split = line.split() #2 and 4 have resi number, -1 has cst number, 1 and 3 have atom
            cst_dict[split[-1]] = (int(split[2]), int(split[4]), split[1], split[3])
    
    #Which structures do we care about (ensemble)
    ens_structs = []
    if not nmr:
        #with open(struct_path+ensemble_file, 'r') as in_file:
        #    for raw_line in in_file:
        #        line = raw_line.strip()
        #        if 'score' not in line:
        #            ens_structs.append(line.split('/')[-1].split('.')[0])
        ens_structs = get_ensemble_list(struct_path+ensemble_file)
    
    #initialize the columns
    #CstFile1_100_score
    #file with score for constraints
    #print(score_file)
    with open(struct_path+score_file, 'r') as in_file:
        first_line = in_file.readline()
        first_line = in_file.readline()
        cols=first_line.split()
    
    cst_eval = {}
    for key in cst_dict.keys():
        cst_eval[key] = []
    
    full_df = pd.read_csv(struct_path+score_file, delim_whitespace=True, skiprows=1)[cols]
    if nmr:
        for index, row in full_df.iterrows():
            for col in cols:
                #if 'score' in col and 'CstFile' in col:
                if 'measure' in col and 'CstFile' in col:
                    #print(col)
                    num = col.split('_')[1]
                    score = float(row[col])
                    cst_eval[num].append(score)
    else:
        for name in ens_structs:
            #for each file in ensemble, get the constraints

            df = full_df.loc[full_df['description'] == name ]

            for col in cols:
                #if 'score' in col and 'CstFile' in col:
                if 'measure' in col and 'CstFile' in col:
                    #print(col)
                    num = col.split('_')[1]
                    score = float(df[col])
                    cst_eval[num].append(score)
    
    for num in cst_eval.keys():
        pair = cst_dict[num] #pair or interacting residues
        i = pair[0] -1
        j = pair[1] -1
        #print(reference)
        ref = reference[int(num)]
        
        #d = (1/N * sum (d)**-6)-1/6
        dist = (np.sum(np.array(cst_eval[num]) ** -6)/ len(cst_eval[num]))**(-1/6)
        #print(num, dist, ref)
        score = abs(ref - dist)
        if i == j:
            if matrix[i][j] == blank:
                matrix[i][j] = 0
            matrix[i][j] += score
        else:
            if matrix[i][j] == blank:
                matrix[i][j] = 0
                matrix[j][i] = 0
            matrix[i][j] += score
            matrix[j][i] += score
    
    #plotting
    plt.figure(figsize=(8, 8))
    cmap=plt.get_cmap('bwr')
    magnitude = max(np.max(matrix), abs(np.min(matrix)))
    
    #set non noe residue pairs
    #matrix[matrix == -1.0] = -1 * magnitude * 1/5
    for i in range(num_resi):
        for j in range(num_resi):
            if matrix[i][j] == blank:
                matrix[i][j] = -1 * magnitude * 1/5
    
    cmap = truncate_colormap(cmap, 0.0, 1.0)
    plt.imshow(matrix, cmap=cmap, vmin=-magnitude, vmax=magnitude)
    plt.xlabel('Residue')
    plt.ylabel('Residue')
    if nmr:
        plt.title(f'PDB Ensemble Averaged Restraint Deviation {pdb}')
    else:
        plt.title(f'Rosetta Ensemble Averaged Restraint Deviation {pdb}')
    cmap_pos = truncate_colormap(cmap, 0.5, 1.0)
    norm = Normalize(vmin=0, vmax=magnitude)
    mappable = mpl.cm.ScalarMappable(norm=norm, cmap=cmap_pos)
    cbar = plt.colorbar(mappable)
    #cbar.set_cmap(cmap_pos)
    cbar.set_label('Sum of NOE Restraint Deviations')
    #plt.clim(0, magnitude)
    #if pdb == '2KTO':
    #    num_resi -= 1
    y_names = [x+1 for x in range(num_resi)]
    y_ticks = [x for x in range(num_resi)]
    x_names = [x+1 for x in range(num_resi)]
    x_ticks = [x for x in range(num_resi)]
    plt.yticks(y_ticks, y_names)
    plt.xticks(x_ticks, x_names)
    plt.show()
```

```python
def ensemble_noe_dev(pdb, nmr=False, structure='helix'):
    if structure == 'helix':
        struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/'
    elif structure == 'unstructured':
        struct_path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/conf_gen/multi_ripp_scan/xml_test/'
    else:
        print(f'{structure} is not a structure option')
        exit()
    cst_file = pdb + '_noe.cst'
    if nmr:
        score_file = pdb + '_cst_nmr.sc'
    else:
        score_file = pdb + '_cst.sc'
    ensemble_file =  pdb + '_ensemble.txt'
    ref_cst = struct_path + pdb + '_noe.tbl'
    reference = parse_reference(ref_cst)
    
    #what are the nmr constraints (residue and atom)?
    cst_dict = {}
    #file with nmr constraints
    with open(struct_path+cst_file, 'r') as in_file:
        for raw_line in in_file:
            line = raw_line.strip()
            split = line.split() #2 and 4 have resi number, -1 has cst number, 1 and 3 have atom
            cst_dict[split[-1]] = (int(split[2]), int(split[4]), split[1], split[3])
    
    #Which structures do we care about (ensemble)
    ens_structs = []
    if not nmr:
        ens_structs = get_ensemble_list(struct_path+ensemble_file)
    
    #initialize the columns
    #CstFile1_100_score
    #file with score for constraints
    #print(score_file)
    with open(struct_path+score_file, 'r') as in_file:
        first_line = in_file.readline()
        first_line = in_file.readline()
        cols=first_line.split()
    
    cst_eval = {}
    for key in cst_dict.keys():
        cst_eval[key] = []
    
    full_df = pd.read_csv(struct_path+score_file, delim_whitespace=True, skiprows=1)[cols]
    if nmr:
        for index, row in full_df.iterrows():
            for col in cols:
                #if 'score' in col and 'CstFile' in col:
                if 'measure' in col and 'CstFile' in col:
                    #print(col)
                    num = col.split('_')[1]
                    score = float(row[col])
                    cst_eval[num].append(score)
    else:
        for name in ens_structs:
            #for each file in ensemble, get the constraints

            df = full_df.loc[full_df['description'] == name ]

            for col in cols:
                #if 'score' in col and 'CstFile' in col:
                if 'measure' in col and 'CstFile' in col:
                    #print(col)
                    num = col.split('_')[1]
                    score = float(df[col])
                    cst_eval[num].append(score)
    
    score = 0
    
    for num in cst_eval.keys():
        #print(reference)
        ref = reference[int(num)]
        
        #d = (1/N * sum (d)**-6)-1/6
        dist = (np.sum(np.array(cst_eval[num]) ** -6)/ len(cst_eval[num]))**(-1/6)
        #print(num, dist, ref)
        score += abs(ref - dist)
    return(score)
```

```python
def get_ens_min_scoring(scorefile, ensemble):
    cols = None
    with open(scorefile, 'r') as in_file:
        first_line = in_file.readline()
        first_line = in_file.readline()
        cols=first_line.split()
    
    full_df = pd.read_csv(scorefile, delim_whitespace=True, skiprows=1)[cols]
    
    scores = {}
    
    #print(ensemble)
    #for name in ensemble:
    #    df = full_df.loc[full_df['description'] == name ]
    for index, row in full_df.iterrows():
        #print()
        if type(row['description']) is not float and row['description'].split('/')[-1].split('.')[0] in ensemble:
            scores[float(row['total_score'])] = row['description']
            
    #return df.nsmallest(n=1, columns=['total_score'])['description'].iloc[0]
    low_score = min (scores.keys())
    return scores[low_score]
```

```python
path = '/home/tydingcw/Documents/EGFR_antibodies/RiPP_design/af_pred/top_pred/'
#file_list = [path + '8CWX_native/bb_8CWX_8_rename_0009.pdb.gz', path + '8CWX/sample_8CWX_0_0494.pdb.gz']

total_score_dict = {}

for pdb in pdbs:
    total_score_dict[pdb] = []
    
    ens_structs = get_ensemble_list(path+pdb+'_ensemble.txt')

    file_list = [path + get_min_scoring(f'{path}{pdb}_bb_relax.sc') + '.pdb.gz', 
                 path + get_ens_min_scoring(f'{path}{pdb}_af_sample.sc', ens_structs) + '.pdb.gz']
    
    #print(file_list)
    
    plot_eng = {}

    for filename in file_list:

        aa_energies = {} #individual aa
        columns = []
        pose = []
        pose_energies = {}
        read = False
        with gzip.open(filename, 'rt') as infile:
            for raw_line in infile:
                line = raw_line.strip()
                if '#BEGIN_POSE_ENERGIES_TABLE' in line:
                    read = True
                elif '#END_POSE_ENERGIES_TABLE' in line:
                    read = False
                elif read:
                    split = line.split()
                    if columns == []:
                        columns = split
                        for x in columns:
                            aa_energies[x] = []
                    elif split[0] not in ['weights', 'pose']:
                        for i in range(len(split)):
                            aa_energies[columns[i]].append(split[i])
                    elif split[0] == 'pose':
                        pose = split
                        for i in range(1,len(pose)):
                            pose_energies[columns[i]] = pose[i]

        #print(pose_energies)
        total_score_dict[pdb].append(float(pose_energies['total']))

    
x_loc = np.arange(len(total_score_dict.keys()))
width = 0.2
fig, ax = plt.subplots(figsize=(3, 4))
plt.bar(x_loc - width, [total_score_dict[pdb][0] for pdb in total_score_dict.keys()], width*2, label='NMR')
plt.bar(x_loc + width, [total_score_dict[pdb][1] for pdb in total_score_dict.keys()], width*2, label='Rosetta')
plt.xlabel('PDB ID')
plt.ylabel('Rosetta Energy Score')
plt.xticks(x_loc, total_score_dict.keys())
plt.legend()
plt.show()
```
